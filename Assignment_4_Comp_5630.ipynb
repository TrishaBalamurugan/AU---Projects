```python
import pandas as pd

# -----------------------------
# LOAD DATA
# -----------------------------
file = "Asssignment4_Data.xlsx"

train_df = pd.read_excel(file, sheet_name="Train")
test_df = pd.read_excel(file, sheet_name="Test")

# -----------------------------
# PRIOR PROBABILITIES
# -----------------------------
P_Apartment = 8/25
P_House = 9/25
P_Condo = 8/25

# -----------------------------
# CONDITIONAL PROBABILITIES
# -----------------------------

# Bathrooms
P_bath_A = {1: 5/8, 1.5: 2/8, 2.5: 1/8}
P_bath_H = {1: 6/9, 1.5: 2/9, 2.5: 1/9}
P_bath_C = {1: 4/8, 1.5: 3/8, 2.5: 1/8}

# Bedrooms
P_bed_A = {3: 6/8, 4: 2/8, 5: 0}
P_bed_H = {3: 6/9, 4: 3/9, 5: 0}
P_bed_C = {3: 6/8, 4: 0, 5: 2/8}

# Garages
P_gar_A = {0: 2/8, 1: 5/8, 1.5: 1/8, 2: 0}
P_gar_H = {0: 0, 1: 4/9, 1.5: 0, 2: 5/9}
P_gar_C = {0: 0, 1: 6/8, 1.5: 0, 2: 2/8}


# -----------------------------
# PREDICTION FUNCTION
# -----------------------------
def predict(bath, bed, gar):

    # Apartment
    pA = P_Apartment
    pA *= P_bath_A.get(bath, 0)
    pA *= P_bed_A.get(bed, 0)
    pA *= P_gar_A.get(gar, 0)

    # House
    pH = P_House
    pH *= P_bath_H.get(bath, 0)
    pH *= P_bed_H.get(bed, 0)
    pH *= P_gar_H.get(gar, 0)

    # Condo
    pC = P_Condo
    pC *= P_bath_C.get(bath, 0)
    pC *= P_bed_C.get(bed, 0)
    pC *= P_gar_C.get(gar, 0)

    # Normalize probabilities
    total = pA + pH + pC
    if total != 0:
        pA /= total
        pH /= total
        pC /= total

    # MAP decision
    prediction = max(
        [("Apartment", pA), ("House", pH), ("Condo", pC)],
        key=lambda x: x[1]
    )[0]

    return pA, pH, pC, prediction


# -----------------------------
# RUN ON TEST DATA
# -----------------------------
print("===== Naïve Bayes Results =====")

for i, row in test_df.iterrows():

    bath = row["Bathrooms"]
    bed = row["# Bedrooms"]
    gar = row["# Garages"]

    pA, pH, pC, pred = predict(bath, bed, gar)

    print(f"\nTest Example {i+1}: ({bath}, {bed}, {gar})")
    print(f"P(Apartment | X) = {pA:.4f}")
    print(f"P(House | X) = {pH:.4f}")
    print(f"P(Condo | X) = {pC:.4f}")
    print(f"Final Prediction (MAP) = {pred}")
```

#Decision Tree Classifier

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# -----------------------------
# LOAD DATA
# -----------------------------
file = "Asssignment4_Data.xlsx"

train_df = pd.read_excel(file, sheet_name="Train")
test_df = pd.read_excel(file, sheet_name="Test")

# -----------------------------
# SPLIT FEATURES AND LABEL
# -----------------------------
X_train = train_df.drop(columns=["House ID", "Construction type"])
y_train = train_df["Construction type"]

X_test = test_df.drop(columns=["House ID", "Construction type"])
y_test = test_df["Construction type"]

# -----------------------------
# 1. DEFAULT DECISION TREE
# -----------------------------
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

train_pred = model.predict(X_train)
test_pred = model.predict(X_test)

train_acc = accuracy_score(y_train, train_pred)
test_acc = accuracy_score(y_test, test_pred)

print("DEFAULT TREE RESULTS")
print("Train Accuracy:", train_acc)
print("Test Accuracy:", test_acc)

# -----------------------------
# 2. MAX DEPTH EXPERIMENT
# -----------------------------
print("\nDEPTH EXPERIMENT RESULTS")

best_depth = 1
best_acc = 0

for d in range(1, 11):
    model = DecisionTreeClassifier(max_depth=d, random_state=42)
    model.fit(X_train, y_train)

    acc = accuracy_score(y_test, model.predict(X_test))
    print(f"Depth {d} -> Test Accuracy: {acc:.4f}")

    if acc > best_acc:
        best_acc = acc
        best_depth = d

print("\nBest Depth:", best_depth)
print("Best Accuracy:", best_acc)

# -----------------------------
# 3. FINAL TREE (BEST DEPTH)
# -----------------------------
final_model = DecisionTreeClassifier(max_depth=best_depth, random_state=42)
final_model.fit(X_train, y_train)

rules = export_text(final_model, feature_names=list(X_train.columns))
print("\nDECISION TREE RULES:\n")
print(rules)
```

### (b) Test Accuracy
When evaluated on the test dataset, the model achieved a **test accuracy of 0.4**.

---

## 2. Effect of Maximum Depth

Different values of maximum depth were tested from 1 to 10.

- At **depth = 1**, the model underfits and gives low accuracy (0.4).
- At **depth = 2**, the model achieves the best performance with a **test accuracy of 0.8**.
- For depths greater than 2, the accuracy decreases again to 0.4 due to overfitting.

### Conclusion:
Restricting the maximum depth improves generalization by reducing overfitting.  
The best depth for this dataset is **max_depth = 2**.
