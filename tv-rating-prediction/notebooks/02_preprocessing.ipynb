# Feature Engineering & Preprocessing

This notebook prepares features for ML models.

```

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```

# Load dataset
df = pd.read_csv("../data/top_rated_2000webseries.csv")
df.head()
```

```

# Basic cleaning
df = df.drop_duplicates()
df = df.dropna(subset=['rating', 'overview', 'genre'])

df.shape
```

```

# Split genres into lists
df['genre_list'] = df['genre'].str.split(', ')
df[['genre', 'genre_list']].head()
```

```

# Multi-hot encode genres
from sklearn.preprocessing import MultiLabelBinarizer

mlb = MultiLabelBinarizer()
genre_encoded = mlb.fit_transform(df['genre_list'])

genre_df = pd.DataFrame(
    genre_encoded,
    columns=mlb.classes_,
    index=df.index
)

genre_df.head()
```

```

# Combine numeric + genre features
numeric_features = df[['popularity', 'votes']].fillna(0)
X_structured = pd.concat([numeric_features, genre_df], axis=1)

X_structured.head()
```

```

# TF-IDF for overview text
tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words='english',
    ngram_range=(1,2)
)

X_text = tfidf.fit_transform(df['overview'])
X_text.shape
```

```

# Target variable
y = df['rating']
```

```

# Train-test split
X_train_struct, X_test_struct, y_train, y_test = train_test_split(
    X_structured, y, test_size=0.2, random_state=42
)

X_train_text, X_test_text, _, _ = train_test_split(
    X_text, y, test_size=0.2, random_state=42
)

X_train_struct.shape, X_train_text.shape
```


## Preprocessing Summary
- Numeric features retained
- Genres multi-hot encoded
- Overview text vectorized with TF-IDF
- Train/test split completed
