# Exploratory Data Analysis – TV Show Rating Prediction

```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
plt.style.use('default')
```

```

df = pd.read_csv("../data/top_rated_2000webseries.csv")
df.head()
```

```

df.shape
```

```

df.info()
```

```

df.isnull().sum().sort_values(ascending=False)
```

```

df['rating'].describe()
```

```

plt.figure()
sns.histplot(df['rating'], bins=20, kde=True)
plt.title("Distribution of TV Show Ratings")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()
```


## Key EDA Insights
- Ratings are tightly clustered
- Popularity and votes show weak correlations
- Genre and language introduce meaningful structure
- Overview text provides rich semantic features
